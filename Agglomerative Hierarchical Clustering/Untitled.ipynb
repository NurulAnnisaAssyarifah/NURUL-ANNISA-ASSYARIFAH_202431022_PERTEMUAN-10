```python
#202431022_Nurul Annisa Assyarifah

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
```

```python
#202431022_Nurul Annisa Assyarifah

df = pd.read_csv("wearable_health_devices_performance_upto_26june2025.csv", sep = ';')
df
```

```python
#202431022_Nurul Annisa Assyarifah

df = np.array(df[['User_Satisfaction_Rating', 'Performance_Score']])
df

#agar lebih mudah aja makanya pake 2 kolom. 
#Karena kedua kolom tsb saling berhubungan, sama -sama membahas kualitas dari dataset yang kita gunakan tadi

#diubah ke array agar lebih mudah di proses
```

```python
#202431022_Nurul Annisa Assyarifah

plt.title("Dendogram Metode Single Linkage")
dendrogram = sch.dendrogram(sch.linkage(df, method = 'single'))
```

```python
#202431022_Nurul Annisa Assyarifah
plt.title("Dendogram Metode Average Linkage")
dendrogram = sch.dendrogram(sch.linkage(df, method = 'average'))

#Lebih seimbang karena ngukur nya berdasarkan rata-rata
```

```python
#202431022_Nurul Annisa Assyarifah

plt.title("Dendogram Metode Complete Linkage")
dendrogram = sch.dendrogram(sch.linkage(df, method = 'complete'))

#Gabungin cluster berdasarkan jarak terjauh
```

```python
#202431022_Nurul Annisa Assyarifah

plt.title("Dendogram Metode Ward Linkage")
dendrogram = sch.dendrogram(sch.linkage(df, method = 'ward'))
```

```python
#202431022_Nurul Annisa Assyarifah

from sklearn.cluster import AgglomerativeClustering as acg
```

```python
#202431022_Nurul Annisa Assyarifah

ac = acg(n_clusters = 4, linkage = 'single') #isinya tuh algoritma clustering tsb, disini kita set tuh 4 cluster.
output_eu = ac.fit_predict(df) #latih data menggunakan algoritma tsb

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow') #jika warna nya sama maka akan di samakan/ dikelommpokkan (eu)
plt.title('Scatter Plot')
plt.xlabel('Customer Satisfication')
plt.ylabel('Performance Score')
plt.show()

#cocok mendeteksi nilai outlier
```

```python
#202431022_Nurul Annisa Assyarifah

ac = acg(n_clusters = 4, linkage = 'average')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter Plot Average Linkage')
plt.xlabel('Customer Satisfication')
plt.ylabel('Performance Score')
plt.show()

#cluster nya lebih merata, disini kita bisa membagi jadi 4 cluster
#ungu terendah
#merah tuh menengah kerendah
#biru tuh menengah keatas
#kuning itu tertinggi
#Btw ga ada cluster anomali
```

```python
#202431022_Nurul Annisa Assyarifah

ac = acg(n_clusters = 4, linkage = 'complete')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter Plot Complete Linkage')
plt.xlabel('Customer Satisfication')
plt.ylabel('Performance Score')
plt.show()

#Disini ga terlalu keliatan karena kita ga menghitung secara langsung. Kalo mau tau lebih jelas ya kaya ibunya hitung manual
```

```python
#202431022_Nurul Annisa Assyarifah

ac = acg(n_clusters = 4, linkage = 'ward')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter Plot Ward Linkage')
plt.xlabel('Customer Satisfication')
plt.ylabel('Performance Score')
plt.show()
```
